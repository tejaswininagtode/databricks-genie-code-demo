# TPC-H Business Insights Generation

## Overview

This notebook generates comprehensive business insights from the enriched TPC-H customer orders data. It analyzes revenue patterns by nation, customer market segments, and order status, storing all insights in the `default.tpch_insights` Delta table for dashboard consumption.

**Key Insights Generated:**
* Revenue analysis by nation
* Customer segment performance
* Order status distribution

**Output Table:** default.tpch_insights

```python
%sql
CREATE OR REPLACE TEMP VIEW customer_orders_enriched AS
SELECT 
  o.o_orderkey,
  o.o_orderdate,
  o.o_totalprice,
  o.o_orderstatus,
  o.o_orderpriority,
  c.c_custkey,
  c.c_name,
  c.c_acctbal,
  c.c_mktsegment,
  n.n_name AS nation_name,
  n.n_regionkey
FROM samples.tpch.orders o
INNER JOIN samples.tpch.customer c ON o.o_custkey = c.c_custkey
INNER JOIN samples.tpch.nation n ON c.c_nationkey = n.n_nationkey;
```

```python
%sql
CREATE TABLE IF NOT EXISTS default.tpch_insights (
  insight_type STRING,
  dimension STRING,
  metric_name STRING,
  metric_value DECIMAL(18,2),
  record_count BIGINT,
  order_date_min DATE,
  order_date_max DATE,
  generated_at TIMESTAMP
) USING DELTA;
```

```python
%sql
DELETE FROM default.tpch_insights WHERE insight_type = 'revenue_by_nation';

INSERT INTO default.tpch_insights
SELECT 
  'revenue_by_nation' AS insight_type,
  nation_name AS dimension,
  'total_revenue' AS metric_name,
  SUM(o_totalprice) AS metric_value,
  COUNT(*) AS record_count,
  MIN(o_orderdate) AS order_date_min,
  MAX(o_orderdate) AS order_date_max,
  CURRENT_TIMESTAMP() AS generated_at
FROM customer_orders_enriched
GROUP BY nation_name;

INSERT INTO default.tpch_insights
SELECT 
  'revenue_by_nation' AS insight_type,
  nation_name AS dimension,
  'avg_order_value' AS metric_name,
  AVG(o_totalprice) AS metric_value,
  COUNT(*) AS record_count,
  MIN(o_orderdate) AS order_date_min,
  MAX(o_orderdate) AS order_date_max,
  CURRENT_TIMESTAMP() AS generated_at
FROM customer_orders_enriched
GROUP BY nation_name;

INSERT INTO default.tpch_insights
SELECT 
  'revenue_by_nation' AS insight_type,
  nation_name AS dimension,
  'customer_count' AS metric_name,
  COUNT(DISTINCT c_custkey) AS metric_value,
  COUNT(*) AS record_count,
  MIN(o_orderdate) AS order_date_min,
  MAX(o_orderdate) AS order_date_max,
  CURRENT_TIMESTAMP() AS generated_at
FROM customer_orders_enriched
GROUP BY nation_name;
```

```python
%sql
DELETE FROM default.tpch_insights WHERE insight_type = 'segment_analysis';

INSERT INTO default.tpch_insights
SELECT 
  'segment_analysis' AS insight_type,
  c_mktsegment AS dimension,
  'total_revenue' AS metric_name,
  SUM(o_totalprice) AS metric_value,
  COUNT(*) AS record_count,
  MIN(o_orderdate) AS order_date_min,
  MAX(o_orderdate) AS order_date_max,
  CURRENT_TIMESTAMP() AS generated_at
FROM customer_orders_enriched
GROUP BY c_mktsegment;

INSERT INTO default.tpch_insights
SELECT 
  'segment_analysis' AS insight_type,
  c_mktsegment AS dimension,
  'avg_order_value' AS metric_name,
  AVG(o_totalprice) AS metric_value,
  COUNT(*) AS record_count,
  MIN(o_orderdate) AS order_date_min,
  MAX(o_orderdate) AS order_date_max,
  CURRENT_TIMESTAMP() AS generated_at
FROM customer_orders_enriched
GROUP BY c_mktsegment;
```

```python
%sql
DELETE FROM default.tpch_insights WHERE insight_type = 'order_status';

INSERT INTO default.tpch_insights
SELECT 
  'order_status' AS insight_type,
  o_orderstatus AS dimension,
  'total_revenue' AS metric_name,
  SUM(o_totalprice) AS metric_value,
  COUNT(*) AS record_count,
  MIN(o_orderdate) AS order_date_min,
  MAX(o_orderdate) AS order_date_max,
  CURRENT_TIMESTAMP() AS generated_at
FROM customer_orders_enriched
GROUP BY o_orderstatus;
```

```python
%sql
SELECT 
  insight_type,
  COUNT(*) as insight_count,
  SUM(record_count) as total_records_analyzed,
  MIN(generated_at) as generated_at
FROM default.tpch_insights
GROUP BY insight_type, generated_at
ORDER BY insight_type;
```
